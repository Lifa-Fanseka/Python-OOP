```python
#Creating the constructor and then stroing the objects with data
#Self is used to refer to the current object, that is why it is important to have it in every constructor

class Student:
    def __init__(self, name, student_number, mark):
        self.name = name
        self.student_number = student_number
        self.mark = mark
        print(f"{self.name}, {self.student_number}, {self.mark}")

s1 = Student("Lifa", 2233333333, 99)
s2 = Student("Fanseka", 2020039338, 89)
```

```python
#Methods

#Methods are functions that are defined inside a class

class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark
    def display_info(self):
        print(f"My name is: {self.name} and my mark is: {self.mark}")
    def passed(self):
        return self.mark>=50

s1 = Student("Lifa", 90)
s2 = Student("Walter", 78)
s3 = Student("Endlo", 20)

s3.display_info()
print(s3.passed())
```

```python
#Encapsulation
#This refers to how the data is accessed

#The difference between public and protected attributes

#PUBLIC
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
#PROTECTED
class BankAccount:
    def __init__(self, owner,balance):
        self.owner= owner
        self._balance = balance #The single underscore indicates that this is for internal use only
```

```python
#Name mangling
#This is designed to avoid accidental clashes more especially in sub classes.
#The difference between it and Protected is the underscore, name mangling uses double underscore

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, amount):
        if amount<0:
            raise ValueError("Amount cannot be negative")
        self._balance=amount

    account = BankAccount("Amina", 1000)
    print(account.balance)
    account.balance=1200 #One can also write this like account.set_balance(1200), both work
    print(account.balance)
```
